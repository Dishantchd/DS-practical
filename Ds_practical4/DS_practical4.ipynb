```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/TeachingRatings.csv")
df
```

```python
import matplotlib.pyplot as plt

avg_eval = df.groupby('division')['eval'].mean()

avg_eval.plot(kind='bar')

plt.title('Average Teaching Evaluation by Division')
plt.ylabel('Average Evaluation Score')
plt.show()
```

Left bar = Lower division

Right bar = Upper division

Lower division bar is slightly taller

Therefore, instructors teaching lower-division courses receive slightly better evaluations on average.

Average evaluation (Lower division) = 4.087

Average evaluation (Upper division) = 3.953

Lower-division instructors receive slightly higher evaluations.

```python
import matplotlib.pyplot as plt

plt.scatter(df['age'], df['eval'])

plt.xlabel('Age')
plt.ylabel('Evaluation Score')
plt.title('Age vs Teaching Evaluation')
plt.show()
```

```python
import matplotlib.pyplot as plt

male = df[df['gender'] == 'male']
female = df[df['gender'] == 'female']

plt.scatter(male['age'], male['eval'], label='Male')
plt.scatter(female['age'], female['eval'], label='Female')

plt.xlabel('Age')
plt.ylabel('Evaluation Score')
plt.title('Age vs Evaluation by Gender')

plt.legend()
plt.show()
```

Scatter plot by gender shows similar patterns for males and females.

No strong age effect

```python
# Scatter Plot by Gender and Tenure
import matplotlib.pyplot as plt

groups = df.groupby(['gender', 'tenure'])

for name, group in groups:
    plt.scatter(group['age'],
                group['eval'],
                label=f'{name[0]}-{name[1]}')

plt.xlabel('Age')
plt.ylabel('Evaluation Score')
plt.title('Age vs Evaluation by Gender and Tenure')

plt.legend()
plt.show()
```

Scatter plot by gender and tenure creates four groups.

Age remains a weak predictor of evaluation scores across all groups
