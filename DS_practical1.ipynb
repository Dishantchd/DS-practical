```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/drive/MyDrive/StudentsPerformance.csv")
df
```

```python
df.info()
```

```python

arr = df[['math score', 'reading score', 'writing score']].to_numpy()

print("Shape:", arr.shape)


print("Size:", arr.size)

print("Dimension:", arr.ndim)
```

```python
# Select specific rows and numerical columns
selected_data = df.loc[0:3, ['math score', 'reading score']]

print(selected_data)
```

```python
arr = selected_data.to_numpy()

# Divide data into two matrices
A = arr[:2, :]
B = arr[2:4, :]

print("Matrix A:\n", A)
print("Matrix B:\n", B)
```

```python
print("Addition:\n", A + B)
```

```python
print("Subtraction:\n", A - B)
```

```python
print("Subtraction:\n", A * B)
```

```python
print("Matrix Product:\n", np.dot(A, B))
```

```python
print("Inverse of A:\n", np.linalg.inv(A))
```

```python
print("Transpose of A:\n", A.T)
```

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include='object').columns

# Apply One-Hot Encoding
encoded_df = pd.get_dummies(
    df,
    columns=categorical_columns,
    dtype=int
)

# Display encoded dataset
print(encoded_df.head())

# Display new shape
print("New Shape:", encoded_df.shape)
```
