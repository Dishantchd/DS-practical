```python
import pandas as pd
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm

df = pd.read_csv("TeachingRatings.csv")

# Regression
model = smf.ols('eval ~ C(gender)', data=df).fit()

print(model.summary())
```

p-value = 0.001 < 0.05

Reject the Null Hypothesis.

Conclusion: Gender significantly affects teaching evaluation scores. Male instructors receive, on average, about 0.168 points higher evaluation scores than female instructors in this dataset.

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 40, 50, 100],
    labels=['Young', 'Middle', 'Old']
)

model2 = smf.ols('beauty ~ C(age_group)', data=df).fit()

anova_table = anova_lm(model2)

print(anova_table)
```

sum_sq = 17.800717

F = 15.189048

p = 4.094085e-07

Since:

p-value = 4.094085e-07 < 0.05

Reject the Null Hypothesis.

Conclusion: Beauty scores differ significantly across age groups.

```python
model3 = smf.ols('eval ~ beauty', data=df).fit()

print(model3.summary())
```

beauty coefficient = 0.1330

p-value = 0.000

R-squared = 0.036

F-statistic = 17.08

Since:

p-value < 0.05

Reject the Null Hypothesis.

Conclusion: There is a significant positive relationship between beauty score and teaching evaluation score.

As beauty score increases by 1 unit, the evaluation score increases by approximately 0.133 points.
