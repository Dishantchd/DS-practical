Teaching Ratings Dataset - Probability & Hypothesis Testing

Q1: P(evaluation score > 4.5)

Q2: P(3.5 < evaluation score < 4.2)

Q3: Two-tailed z-test - basketball team vs regional players


```python
import pandas as pd
from scipy import stats
import math
```

```python
df = pd.read_csv("TeachingRatings.csv")
df
```

```python
mean_eval = df['eval'].mean()
std_eval = df['eval'].std(ddof=1)

print(f"Mean evaluation score : {mean_eval:.4f}")
print(f"Std dev (sample)      : {std_eval:.4f}")
```

Q1: P(evaluation score > 4.5)**bold text**

```python
# Normal-distribution (theoretical) probability
z1 = (4.5 - mean_eval) / std_eval
p_theoretical = 1 - stats.norm.cdf(z1)
print(f"  z-score              : {z1:.4f}")
print(f"  P(X > 4.5) [normal]  : {p_theoretical:.4f} -> {p_theoretical*100:.2f}%")
```

```python
# Actual proportion in the real data (for comparison)
p_actual = (df['eval'] > 4.5).mean()
print(f"  P(X > 4.5) [actual]  : {p_actual:.4f} -> {p_actual*100:.2f}%")
```

**Q2: P(3.5 < evaluation score < 4.2)
**



```python

z_low = (3.5 - mean_eval) / std_eval
z_high = (4.2 - mean_eval) / std_eval
p_theoretical2 = stats.norm.cdf(z_high) - stats.norm.cdf(z_low)
print(f"  z_low, z_high        : {z_low:.4f}, {z_high:.4f}")
print(f"  P(3.5<X<4.2)[normal] : {p_theoretical2:.4f} -> {p_theoretical2*100:.2f}%")

p_actual2 = ((df['eval'] > 3.5) & (df['eval'] < 4.2)).mean()
print(f"  P(3.5<X<4.2)[actual] : {p_actual2:.4f} -> {p_actual2*100:.2f}%")
```

Q3: Two-tailed z-test (basketball scores

```python
mu0 = 12          # historic pro mean
sigma = 5.5        # known population std dev
n = 36             # sample size
x_bar = 10.7        # sample mean
```

```python
se = sigma / math.sqrt(n)
z_stat = (x_bar - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))
alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha / 2)

print(f"  Standard error       : {se:.4f}")
print(f"  z statistic          : {z_stat:.4f}")
print(f"  Two-tailed p-value   : {p_value:.4f}")
print(f"  Critical z (a=0.05)  : +/-{z_critical:.4f}")

if abs(z_stat) > z_critical:
    decision = "REJECT H0 - means are significantly different"
else:
    decision = "FAIL TO REJECT H0 - no significant difference found"
print(f"  Decision             : {decision}")
```
